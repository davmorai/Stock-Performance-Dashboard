```python
import plotly.express as px
import pandas as pd

# 1. Daten laden
df = pd.read_csv("stock_sector.csv")

# 2. Treemap erstellen
fig = px.treemap(
    df,
    path=[px.Constant(""), "Sektor"], 
    values="Weight",                               
    hover_name="Ticker",                           
    color="Performance %",                         
    color_continuous_scale="RdYlGn",               
    color_continuous_midpoint=0                    
)

# Layout anpassen
fig.update_layout(
    margin=dict(t=50, l=25, r=25, b=25),
    width=800,  # Breite im Notebook optional anpassen
    height=500  # Höhe im Notebook optional anpassen
)
# Diagramm anzeigen
fig.show()
```
