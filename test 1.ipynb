```python
import streamlit as st
import yfinance as yf
import plotly.express as px
import pandas as pd

# 1. Seiten-Konfiguration (Macht es schön breit und gibt einen Titel)
st.set_page_config(page_title="Mein Aktien Dashboard", layout="wide")
st.title("📈 Mein einfaches Aktien-Dashboard")

# 2. Interaktives Dropdown-Menü erstellen
# st.selectbox ist das Streamlit-Element für Dropdowns
ticker_symbol = st.selectbox(
    "Wähle eine Aktie aus:",
    ("AAPL", "MSFT", "GOOGL", "TSLA", "NVDA")
)

# 3. Daten abrufen (API Call zu Yahoo Finance)
# Wir holen die Daten der letzten 30 Tage mit einem 15-Minuten-Intervall
@st.cache_data # Das speichert die Daten zwischen, damit die App schneller lädt
def load_data(ticker):
    stock_data = yf.Ticker(ticker)
    # history() zieht die Preisdaten. Period: 1mo (1 Monat), interval: 15m (15 Minuten)
    df = stock_data.history(period="1mo", interval="15m")
    # Den Index (die Zeitstempel) als normale Spalte hinzufügen, Plotly mag das lieber
    df = df.reset_index()
    return df

data = load_data(ticker_symbol)

# 4. Daten visualisieren mit Plotly Express
if not data.empty:
    st.subheader(f"Preisverlauf für {ticker_symbol}")
    
    # Der Plotly Express Aufruf: extrem simpel!
    # x ist die Zeit (Datetime), y ist der Schlusskurs (Close)
    fig = px.line(
        data, 
        x="Datetime", 
        y="Close", 
        title=f"{ticker_symbol} - Letzter Monat (15-Minuten Intervalle)"
    )
    
    # Ein bisschen Styling, um näher an dein Dark-Theme Bild zu kommen
    fig.update_layout(
        template="plotly_dark", # Dark Mode!
        xaxis_title="Datum / Uhrzeit",
        yaxis_title="Preis in USD"
    )
    
    # 5. Den Graphen im Dashboard anzeigen
    st.plotly_chart(fig, use_container_width=True)
else:
    st.error("Keine Daten für dieses Symbol gefunden.")
```
